problem: Let $n \in \mathbb N$. If $n^2$ is even, then $n$ is even.

Let $n^2 = (2k+1)^2$, then:
$$ n^2 = 4k^2 + 4k + 1$$
$$ n^2 = 2(2k^2 + 2k) + 1 $$
Assume $p = 2k^2 + 2k \in \mathbb Z$, then:
$$ n^2 = 2p + 1 $$
Therefore, if $n^2$ is even, then $n$ is even. $\blacksquare$

```python
def is_even(n) -> bool:
    return n % 2 == 0

def verify_theorem_04(limit: int = 5000) -> bool:
    for n in range(1, limit):
        if is_even(n*n) and not is_even(n): return False

    return True

print(f'Exercise 04: {'PASS' if verify_theorem_04() else "FAIL"}')
```
