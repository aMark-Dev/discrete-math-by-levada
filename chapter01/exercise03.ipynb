Problem: Let $p \in \mathbb N$ a prime number, such that $p$ is a factor of $ab$. Then $p$ is a factor of $a$ or $b$. For $a=b$, we have witch, if $p$ divide $a^2$, then $p$ divide $a$.

If $ p | a^2 \Leftrightarrow p | a \times a \Leftrightarrow p | a or p | a$. Therefore $p | a. \blacksquare$

```python
def is_divisible(p,a) -> bool:
    return a % p == 0

def prime_generation(quantity):
    p_list = [2]
    k = p_list[-1]+1

    while(len(p_list) < quantity ):
        is_prime = True
        for p in p_list:

            if k%p == 0:
                break
            if p**2 > k:
                p_list.append(k)
                break
        k+=1

    return p_list

def division(limit: int = 500) -> bool:

    prime_list = prime_generation(limit)
    for p in prime_list:
        for a in range(0, limit**2):
            if is_divisible(p,a*a) and not is_divisible(p,a): return False
    return True

print(f'Exercise 03: {"PASS" if division() else "FAIL"}')
```
