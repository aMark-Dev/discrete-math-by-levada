Problem: $ \forall x,y \in \mathbb{R} $ if $x$ and $y$ are rational, then $x+y$ is rational.

Let $\forall x,y \in \mathbb R:x=\frac{a}{b},y=\frac{c}{d}, b,d\neq 0$, for any $a,b,c,d \in \mathbb Z$, then:
$$ x+y = \frac{a}{b} + \frac{c}{d} $$
$$  = \frac{ad+cb}{db} $$
Assume $ p = ad+cb, q = db \in \mathbb Z$, then:
$$ x+y = \frac{p}{q} $$
If $p,q$ is integer and $q \neq 0$, then $x+y$ is rational. $\blacksquare$

```python
def is_integer(p,q) -> bool:
    bool_value_p = type(p) == int
    bool_value_q = type(q) == int

    return bool(bool_value_p*bool_value_q)

def rational_sum(limit: int = 50) -> bool:

    is_rational = True

    for a in range(-limit, limit):
        for b in range(-limit, limit):
            if b == 0: continue
            for c in range(-limit, limit):
                for d in range(-limit, limit):
                    if d == 0: continue

                    p = a*d + c*b
                    q = d*b

                    if not is_integer(p,q): is_rational= False

    return is_rational

print(f'Exercise 02: {'PASS' if rational_sum() else 'FAIL'}')
        
```
