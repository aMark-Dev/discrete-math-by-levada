Problem: For all $ m,n \in \mathbb{N} $, if $ m $ e $ n $ are even, then $ m+n $ is even.

Assume $ m = 2k $ and $ n = 2w$, for any $k,w \in \mathbb N $. $$ m+n = 2k + 2w = 2 (k+w) $$
Assume $$ v = k+w \in \mathbb{N} \Leftrightarrow m+n = 2v. $$ Therefore the expression $m+n$ is even if $m,n$ are even. $\blacksquare$

```python
def is_even(the_sum) -> bool:
    """Return True if n is even."""

    return the_sum % 2 == 0

def even_sum(limit: int = 10000) -> bool:
    """
    Verify: for all even m, n in [0, limit), m + n is even.
    """
    for m in range(0, limit, 2):        # m even: 0, 2, 4, ...
        for n in range(0, limit, 2):    # n even: 0, 2, 4, ...
            if not(is_even(m+n)):
                return False

    return True

# Run automatically
print (f'Exercise 01: {'PASS' if even_sum() else 'FAIL'}')
```
